- header, data  (level[2])

In the header you can find information like: the exposure time, the telescope, the data of the observations, the size of the date, the observer and sometimes also the informations of the part of the sky we observed 

To access a header, we need to say which extension we want the header for and then we can use the `header` attribute. Header are made up of three pieces: a keyword, a value, and a comment.

# Photometry and Flux: Measuring how bright something is

We want to measure how much light is emitted by each star. When light hits a camera, the photons are converted to electrons which are counted by the camera. The number in each pixel (analog-to-digital units; ADU) is proportional to the number of electrons and therefore photons that hit that pixel. A single star is made up of multiple pixels - so to measure the brightness of a star, we need to add up how many ADU are in all of the pixels associated with the star. Let's take a closer look at the 10th star in the list by plotting just a small area around it. We will start by creating a 2D array (recall we do this with the syntax `array[a:b, c:d]`). We will take 20 pixels in either direction from the center of star 10. This is sometimes called a cutout

Now let's plot star 10. We are going to be a little more explicit about defining first a figure and axis object and then plotting onto the axis. This is a good practice in python. It also allows us to manipulate the figure and axis object whenever we need to. In the past when we did `plt.plot` the figure and axis object were created automatically, but we hadn't assigned them to any variables, so we couldn't do anything with them later.

We will use `plt.subplots(ncol, nrow)` which defines a figure object and an axis or set of axes.

How do we know when we have all of the light from the star? We can define a circle and add up all of ADU in that circle. If we do this for a bunch of different radii, we can see when we stop adding in more ADU. This tells us we've reached the edge of our star.

`sep.sum_circle` is a function that adds up all of the ADU in a circle and using a conversion factor called *gain* to convert from ADU to number of photons. We call the number of photons the flux (energy per unit area per unit time). Let's use it for different circles with radii from 1-20 pixels centered on our star.
